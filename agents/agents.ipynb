```python
import os
from dotenv import load_dotenv
from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_community.vectorstores import Chroma

loader = WebBaseLoader("https://docs.langchain.com/langsmith/observability")
text_document = loader.load()

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)
documents = splitter.split_documents(text_document)

load_dotenv()
embedding = GoogleGenerativeAIEmbeddings(model = "gemini-embedding-001")
vector_db = Chroma.from_documents(documents[:20],embedding=embedding)

retriever = vector_db.as_retriever()
retriever
```

```python
from langchain_classic.tools.retriever import create_retriever_tool
retriever_tool = create_retriever_tool(retriever,"lang-smith-search","search for information about Langsmith.For any question.")
```

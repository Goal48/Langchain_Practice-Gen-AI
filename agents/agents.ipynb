```python
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper
```

```python
api_wrapper=WikipediaAPIWrapper(top_k_results=1,doc_content_chars_max=200)
wiki = WikipediaQueryRun(api_wrapper=api_wrapper)
wiki
```

```python
import os
from dotenv import load_dotenv
from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_community.vectorstores import Chroma

loader = WebBaseLoader("https://docs.langchain.com/langsmith/observability")
text_document = loader.load()

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)
documents = splitter.split_documents(text_document)

load_dotenv()
embedding = GoogleGenerativeAIEmbeddings(model = "gemini-embedding-001")
vector_db = Chroma.from_documents(documents[:20],embedding=embedding)

retriever = vector_db.as_retriever()
retriever
```

```python
from langchain_classic.tools.retriever import create_retriever_tool
retriever_tool = create_retriever_tool(retriever,"lang-smith-search","search for information about Langsmith.For any question.")
```

```python
retriever_tool.name
```

```python
from langchain_community.utilities import ArxivAPIWrapper
from langchain_community.tools import ArxivQueryRun

arxiv_wrapper=ArxivAPIWrapper(top_k_results=1,doc_content_chars_max=200)
arxiv = ArxivQueryRun(api_wrapper = arxiv_wrapper)
arxiv.name
```

```python
tools = [wiki,arxiv,retriever_tool]
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
import langchain
print(langchain.__version__)
```

```python
## Agents
import langchainhub as hub
from langchain_classic.agents import AgentExecutor, create_tool_calling_agent

prompt = hub.pull("hwchase17/openai-tools-agent")

# 4. Create the generic tool calling agent (Works seamlessly with Gemini)
agent = create_tool_calling_agent(llm, tools, prompt)

# 5. Combine the agent with its execution environment
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)

# 6. Run it!
response = agent_executor.invoke({"input": "What is 25 multiplied by 4?"})
print(response["output"])
```
